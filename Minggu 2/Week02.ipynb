### 1. Perbedaan `cv2_imshow()` dan `plt.imshow()` tanpa konversi warna

OpenCV menyimpan warna dengan urutan BGR. `cv2_imshow()` membacanya sebagai BGR sehingga warna terlihat benar. Matplotlib menganggap array tersebut RGB, sehingga kanal merah dan biru terlihat tertukar. Nilai pikselnya tetap sama karena gambar yang dipakai berasal dari array yang sama.

````python
img_bebas = cv.imread('lanskap.jpg')

cv2_imshow(img_bebas)
plt.imshow(img_bebas)
plt.show()

y = img_bebas.shape[0] // 4
x = img_bebas.shape[1] // 4

print('Piksel untuk cv2_imshow:', img_bebas[y, x])
print('Piksel untuk plt.imshow:', img_bebas[y, x])
````

### 2. Perbandingan kanvas `int16` dan `int32`

Tampilannya sama karena semua piksel bernilai 0. Memori `int32` lebih besar karena setiap nilainya memakai 4 byte, sedangkan `int16` memakai 2 byte.

````python
kanvas_int16 = np.zeros(shape=(530,530,3), dtype=np.int16)
kanvas_int32 = np.zeros(shape=(530,530,3), dtype=np.int32)

print('int16 dtype:', kanvas_int16.dtype)
print('int16 min:', kanvas_int16.min())
print('int16 max:', kanvas_int16.max())
print('int16 nbytes:', kanvas_int16.nbytes)

print('int32 dtype:', kanvas_int32.dtype)
print('int32 min:', kanvas_int32.min())
print('int32 max:', kanvas_int32.max())
print('int32 nbytes:', kanvas_int32.nbytes)

plt.subplot(1,2,1)
plt.imshow(kanvas_int16)
plt.title('int16')

plt.subplot(1,2,2)
plt.imshow(kanvas_int32)
plt.title('int32')

plt.show()
````

Berapa persen piksel pada foto KTM yang lebih gelap dari nilai 100?

### Rancangan program

Foto diubah menjadi grayscale. Jumlah piksel dengan nilai di bawah 100 dibagi dengan seluruh piksel, lalu dikali 100 persen.

### Data hasil dan kesimpulan

Nilai dan kesimpulan dicetak oleh program berikut.

````python
img_analisa = cv.imread('ktp.jpg')
gray_analisa = cv.cvtColor(img_analisa, cv.COLOR_BGR2GRAY)

jumlah_gelap = np.sum(gray_analisa < 100)
jumlah_piksel = gray_analisa.size
persentase_gelap = jumlah_gelap / jumlah_piksel * 100

print('Jumlah piksel gelap:', jumlah_gelap)
print('Jumlah seluruh piksel:', jumlah_piksel)
print(f'Persentase piksel gelap: {persentase_gelap:.2f}%')
print(f'Kesimpulan: {persentase_gelap:.2f}% piksel pada foto KTM memiliki nilai di bawah 100.')
````

### Studi Kasus 1: Label identitas yang menyesuaikan ukuran foto

Fungsi yang sama dipakai pada foto lanskap dan potret. Tinggi bar, posisi teks, ukuran huruf, dan ketebalan dihitung dari ukuran gambar.

````python
def tambah_label(citra, teks):
    hasil = citra.copy()
    h, w = hasil.shape[:2]
    tinggi_bar = max(1, h // 12)
    ukuran_huruf = max(0.3, h / 600)
    ketebalan = max(1, h // 300)
    posisi_teks = (tinggi_bar // 5, h - tinggi_bar // 4)

    cv.rectangle(hasil, (0, h - tinggi_bar), (w, h), (0,0,0), -1)
    cv.putText(hasil, teks, posisi_teks, cv.FONT_HERSHEY_SIMPLEX, ukuran_huruf, (255,255,255), ketebalan, cv.LINE_AA)
    return hasil


foto_lanskap = cv.imread('lanskap.jpg')
foto_potret = cv.imread('potret.jpg')

hasil_lanskap = tambah_label(foto_lanskap, 'Dokumentasi PCVK - 2026')
hasil_potret = tambah_label(foto_potret, 'Dokumentasi PCVK - 2026')

plt.figure(figsize=(12,10))

plt.subplot(2,2,1)
plt.imshow(cv.cvtColor(foto_lanskap, cv.COLOR_BGR2RGB))
plt.title(f'Masukan {foto_lanskap.shape[1]} x {foto_lanskap.shape[0]} px')

plt.subplot(2,2,2)
plt.imshow(cv.cvtColor(hasil_lanskap, cv.COLOR_BGR2RGB))
plt.title(f'Keluaran: bar = {foto_lanskap.shape[0]//12} px')

plt.subplot(2,2,3)
plt.imshow(cv.cvtColor(foto_potret, cv.COLOR_BGR2RGB))
plt.title(f'Masukan {foto_potret.shape[1]} x {foto_potret.shape[0]} px')

plt.subplot(2,2,4)
plt.imshow(cv.cvtColor(hasil_potret, cv.COLOR_BGR2RGB))
plt.title(f'Keluaran: bar = {foto_potret.shape[0]//12} px')

plt.tight_layout()
plt.show()
````

### Studi Kasus 2: Thumbnail persegi 1:1

Foto diperkecil tanpa mengubah rasio. Foto diletakkan di tengah kanvas persegi berwarna. Ukuran logo mengikuti ukuran kanvas.

````python
img_produk = cv.imread('produk.jpg')

h, w = img_produk.shape[:2]
sisi = min(h, w)

if w >= h:
    lebar_baru = sisi
    tinggi_baru = max(1, h * sisi // w)
else:
    tinggi_baru = sisi
    lebar_baru = max(1, w * sisi // h)

produk_resize = cv.resize(img_produk, (lebar_baru, tinggi_baru))
thumbnail = np.full((sisi, sisi, 3), (230,240,245), dtype=np.uint8)

x = (sisi - lebar_baru) // 2
y = (sisi - tinggi_baru) // 2
thumbnail[y:y+tinggi_baru, x:x+lebar_baru] = produk_resize

tinggi_logo = max(1, sisi // 9)
ketebalan = max(1, sisi // 300)
ukuran_huruf = max(0.3, sisi / 700)

cv.rectangle(thumbnail, (0,0), (sisi,tinggi_logo), (30,90,220), -1)
cv.putText(thumbnail, 'TOKO SAHRUL', (tinggi_logo//5, 3*tinggi_logo//4), cv.FONT_HERSHEY_SIMPLEX, ukuran_huruf, (255,255,255), ketebalan, cv.LINE_AA)

print('Ukuran foto awal:', w, 'x', h)
print('Ukuran foto setelah resize:', lebar_baru, 'x', tinggi_baru)
print('Ukuran thumbnail:', thumbnail.shape[1], 'x', thumbnail.shape[0])

plt.figure(figsize=(8,8))
plt.imshow(cv.cvtColor(thumbnail, cv.COLOR_BGR2RGB))
plt.title('Thumbnail Produk 1:1')
plt.axis('off')
plt.show()
````
