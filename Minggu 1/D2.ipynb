# D2 - Pengenalan Python untuk Membaca dan Menampilkan Image

## Langkah (Step) 1 - Import Library

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

## Langkah (Step) 2 - Baca Image dari URL

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://commons.wikimedia.org/wiki/Special:FilePath/Cat03.jpg?width=800"]

for url in urls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n')
```

## Langkah (Step) 3 - Melihat Ukuran File Image

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

## Langkah (Step) 4 - Mengakses Pixel

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range (lebar):
    image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

## 2. Kegunaan library
- NumPy digunakan untuk mengolah array dan pixel.
- Pandas digunakan untuk mengolah data tabel.
- OpenCV digunakan untuk membaca dan mengolah gambar.

## 3. Kegunaan resize
Kode tersebut mengecilkan lebar dan tinggi gambar menjadi setengah. Jika tidak dijalankan, gambar tetap memakai ukuran asli.

## 4. Kegunaan `[255,255,255]`
Nilai `[255,255,255]` memberi warna putih pada pixel karena semua channel bernilai 255.

## 5. Keterkaitan pixel dan resolusi
Resolusi adalah jumlah pixel pada lebar dan tinggi gambar. Semakin banyak pixel, gambar semakin jelas dan tidak mudah pecah saat diperbesar.

# Tugas Praktikum D2

## Tugas 2 - Garis Vertikal dan Garis Menyilang Diagonal

```python
image_tugas2 = image_2.copy()
tinggi, lebar = image_tugas2.shape[:2]

for y in range(tinggi):
    image_tugas2[y, int(lebar/2)] = [255,255,255]

for i in range(min(tinggi, lebar)):
    image_tugas2[i, i] = [255,255,255]
    image_tugas2[i, lebar-i-1] = [255,255,255]

final_frame = cv.hconcat((image_2, image_tugas2))
cv2_imshow(final_frame)
```

## Tugas 3 - Garis Horizontal Putih dengan Panjang Tertentu

```python
image_tugas3 = image_2.copy()
tinggi, lebar = image_tugas3.shape[:2]

for y in range(int(lebar/4), int(3*lebar/4)):
    image_tugas3[int(tinggi/2), y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_tugas3))
cv2_imshow(final_frame)
```

## Tugas 4 - Kotak Putih Menggunakan Kumpulan Pixel

```python
image_tugas4 = image_2.copy()
tinggi, lebar = image_tugas4.shape[:2]

for x in range(int(tinggi/3), int(2*tinggi/3)):
    for y in range(int(lebar/3), int(2*lebar/3)):
        image_tugas4[x, y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_tugas4))
cv2_imshow(final_frame)
```
