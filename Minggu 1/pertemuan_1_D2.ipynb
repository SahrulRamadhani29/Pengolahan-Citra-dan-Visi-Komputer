## Langkah (Step) 1 - Import Library

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

## Langkah (Step) 2 - Baca Image dari URL

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://placekitten.com/800/571"]

# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                         #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)  #convert color to RGB
  final_frame = cv.hconcat((image, image_2))       #concatenate image
  cv2_imshow(final_frame)                          #show image
  print('\n')
```

## Langkah (Step) 3 - Melihat Ukuran File Image

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

## Langkah (Step) 4 - Mengakses Pixel

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
    image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

# Pertanyaan Praktikum D2

## 1. Mengapa eksekusi Python menggunakan Google Colab?
Google Colab dipakai karena kode Python dapat langsung dijalankan melalui browser. Kita tidak perlu memasang Python dan semua library secara manual di laptop. Colab juga terhubung dengan Google Drive dan GitHub, jadi notebook lebih mudah disimpan, dibuka kembali, dan dibagikan.

## 2. Kegunaan library
Setiap library mempunyai kegunaan yang berbeda dalam praktikum ini:
- NumPy dipakai untuk mengolah array numerik. Pada pengolahan citra, nilai piksel juga disimpan dalam bentuk array.
- Pandas biasanya digunakan untuk data berbentuk tabel.
- OpenCV atau `cv2` digunakan untuk membaca, mengubah ukuran, mengonversi warna, dan memproses citra.
- `cv2_imshow` digunakan untuk menampilkan citra OpenCV di Google Colab.
- `skimage.io` digunakan untuk membaca citra, termasuk citra yang berasal dari URL.
- `skimage.transform` menyediakan fungsi transformasi citra.
- PIL dapat digunakan untuk membuka dan memanipulasi file gambar.
- Matplotlib digunakan untuk menampilkan citra dalam bentuk plot.
Tidak semua library tersebut harus digunakan pada setiap langkah. Library yang dipakai cukup disesuaikan dengan operasi yang sedang dikerjakan.

## 3. Kegunaan resize
Kode `cv.resize(image, (0,0), fx=0.5, fy=0.5)` mengecilkan lebar dan tinggi gambar menjadi setengah dari ukuran awal. Jika baris ini tidak dijalankan, gambar tetap menggunakan ukuran aslinya. Akibatnya, gambar membutuhkan ruang tampilan dan proses pengolahan yang lebih besar.

## 4. Kegunaan `[255,255,255]`
Nilai `[255,255,255]` digunakan untuk memberi warna putih pada piksel. Angka 255 merupakan nilai intensitas maksimum pada masing-masing channel. Karena ketiga channel mempunyai nilai maksimum yang sama, warna yang dihasilkan adalah putih.

## 5. Keterkaitan pixel dan resolusi
Resolusi gambar ditentukan oleh jumlah piksel pada bagian lebar dan tingginya. Gambar beresolusi tinggi mempunyai lebih banyak piksel sehingga detail yang tersimpan biasanya lebih jelas. Sebaliknya, gambar beresolusi rendah memiliki lebih sedikit piksel dan akan lebih mudah terlihat pecah ketika diperbesar.

## Tugas 2 - Garis Vertikal dan Garis Menyilang Diagonal

```python
image_tugas2 = image_2.copy()
tinggi, lebar = image_tugas2.shape[:2]

for y in range(tinggi):
    image_tugas2[y, int(lebar/2)] = [255,255,255]

for i in range(min(tinggi, lebar)):
    image_tugas2[i, i] = [255,255,255]
    image_tugas2[i, lebar-i-1] = [255,255,255]

final_frame = cv.hconcat((image_2, image_tugas2))
cv2_imshow(final_frame)
```

## Tugas 3 - Garis Horizontal Putih dengan Panjang Tertentu

```python
image_tugas3 = image_2.copy()
tinggi, lebar = image_tugas3.shape[:2]

for y in range(int(lebar/4), int(3*lebar/4)):
    image_tugas3[int(tinggi/2), y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_tugas3))
cv2_imshow(final_frame)
```

## Tugas 4 - Kotak Putih Menggunakan Kumpulan Pixel

```python
image_tugas4 = image_2.copy()
tinggi, lebar = image_tugas4.shape[:2]

for x in range(int(tinggi/3), int(2*tinggi/3)):
    for y in range(int(lebar/3), int(2*lebar/3)):
        image_tugas4[x, y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_tugas4))
cv2_imshow(final_frame)
```
