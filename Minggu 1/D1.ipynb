# D1 - Pengenalan GitHub dan Google Colab

## Langkah praktikum

1. Buat akun GitHub dan lakukan verifikasi email.
2. Buat repository untuk praktikum PCVK.
3. Buka [Google Colab](https://colab.research.google.com/).
4. Pilih menu GitHub pada halaman awal Colab.
5. Hubungkan Colab dengan akun GitHub. Jika repository private, aktifkan akses repository pribadi.
6. Pilih repository dan buat notebook baru.
7. Simpan notebook melalui menu `File > Save a copy in GitHub`.
8. Buka notebook dari GitHub menggunakan tombol `Open in Colab`.
